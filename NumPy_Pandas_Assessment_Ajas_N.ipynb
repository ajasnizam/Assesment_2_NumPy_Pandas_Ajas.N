```python
#Q13

df['Average_Technical_Score'] = df[['Python_Score', 'SQL_Score', 'Pandas_Score', 'NumPy_Score']].mean(axis=1)

# Display the first few rows to verify the new column
df[['Average_Technical_Score']].head()

# Define the conditions based on the rules
conditions = [(df['Average_Technical_Score'] >= 90),(df['Average_Technical_Score'] >= 75),(df['Average_Technical_Score'] >= 60),(df['Average_Technical_Score'] < 60)]

# Define the corresponding performance labels
choices = ['Excellent', 'Good', 'Average', 'Needs Improvement']

# Apply the conditions to create the new column (using 'Unknown' as default to match string data type)
df['Performance'] = np.select(conditions, choices, default='Unknown')

# Display the resulting DataFrame
df
```

```python
#Q14

#Grouping by Department and calculating statistics on Final_Score
dept_summary = df.groupby('Department')['Final_Score'].agg(Number_of_Students='count',Average_Final_Score='mean',Maximum_Final_Score='max',Minimum_Final_Score='min').reset_index()

#Sorting by Average Final Score in descending order
dept_summary_sorted = dept_summary.sort_values(by='Average_Final_Score', ascending=False)

#Displaying the resulting DataFrame
display(dept_summary_sorted)
```

```python
#Q15
import numpy as np

#Convert the Final_Score column into a NumPy array.
score = df['Final_Score'].dropna().to_numpy()

#Calculate using NumPy:
mean_score = np.mean(score)
median_score = np.median(score)
std_dev = np.std(score)
pct_25 = np.percentile(score, 25)
pct_75 = np.percentile(score, 75)
#mean
print(f"Mean Final Score: {mean_score:0.2f}")
#median
print(f"Median Final Score: {median_score:0.2f}")
#standard deviation
print(f"Standard Deviation: {std_dev:0.2f}")
#25th percentile
print(f"25th Percentile: {pct_25:0.2f}")
#75th percentile
print(f"75th Percentile: {pct_75:0.2f}")

#Identify all students whose Final_Score is above the overall mean.
above_mean_students = df[df['Final_Score'] > mean_score]

print(f"\nNumber of students with Final_Score above the overall mean ({mean_score:.2f}): {len(above_mean_students)}")
display(above_mean_students)
```
